```python
import matplotlib.pyplot as plt
import math
```

```python
def exp_lr(epoch):
   init_lr = 1e-5
   k = 0.9
   lr = init_lr * math.exp(-k*epoch)
   return lr

def linear_lr(epoch):
    init_lr=1e-5
    lr=init_lr/(1+.5*epoch)
    return lr
```

```python
plt.plot(range(1,100),list(map(exp_lr,range(1,100))),label='exp')
plt.plot(range(1,100),list(map(linear_lr,range(1,100))),label='linear')
plt.legend()
```
